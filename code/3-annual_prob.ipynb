## Step 2: Get Landsat Images from 2000-2020 for Classification

```python
import ee
# ee.Authenticate()
ee.Initialize(project="ee-joshisur231")
import geemap
import config
import utils
Map = geemap.Map()
```

```python
ic_2000_2012 = utils.get_processed_landsat_collection("LANDSAT/LT05/C02/T1_L2", config.ROI, ['2000-01-01', '2012-12-31'], utils.mask_clouds_landsat75, utils.apply_scale_factors)\
    .map(lambda image: image.select(config.L75_ORIGINAL_BAND_NAMES).rename(config.L75_NEW_BAND_NAMES))
```

```python
ic_2013_2020 = utils.get_processed_landsat_collection("LANDSAT/LC08/C02/T1_L2", config.ROI, ['2013-01-01', '2020-12-31'], utils.mask_clouds_landsat8, utils.apply_scale_factors)\
    .map(lambda image: image.select(config.L8_ORIGINAL_BAND_NAMES).rename(config.L8_NEW_BAND_NAMES))
```

```python
ic_2000_2012.sort("system:start_time")
```

```python
ic_2013_2020.sort("system:start_time")
```

```python
vis_params = {"min":0, "max":0.3, "bands": ["r", "g", "b"]} 
Map.addLayer(ic_2013_2020.median(), vis_params, "ic2000_2012")
Map
```
